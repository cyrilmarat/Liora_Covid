```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import fnmatch
import os
import seaborn as sns

workingpath="/Users/antoine/Desktop/Liora/Projet"
normal_files="/COVID-19_Radiography_Dataset/Normal"
covid_files="/COVID-19_Radiography_Dataset/COVID"
lungOpacity_files="/COVID-19_Radiography_Dataset/Lung_Opacity"
viralPneumonia_files="/COVID-19_Radiography_Dataset/Viral Pneumonia"

size_img=299
```

```python
nb_files_Normal=len(fnmatch.filter(os.listdir(workingpath+normal_files+"/images/"), "*.png"))
all_img_Normal = np.zeros((size_img, size_img, nb_files_Normal),  dtype=np.uint8)
for i in range(0, nb_files_Normal):
    img_Normal = cv2.imread(workingpath+normal_files +"/images/Normal-"+str(i+1)+".png")
    img_Normal= cv2.cvtColor(img_Normal, cv2.COLOR_BGR2GRAY)
    #img_Normal_mask = cv2.imread(workingpath+normal_files +"/masks/Normal-"+str(i+1)+".png", cv2.IMREAD_GRAYSCALE) # pour lire

    #img_mask_resized = cv2.resize(img_Normal_mask, (img_Normal.shape[0], img_Normal.shape[1]), 0, 0, cv2.INTER_NEAREST)
    #img_Normal_and_mask = cv2.bitwise_and(img_Normal,img_mask_resized)
    #img_Normal_and_mask_resized = cv2.resize(img_Normal_and_mask, (size_img, size_img), 0, 0, cv2.INTER_NEAREST)
    all_img_Normal[:,:,i] = img_Normal
```

```python
nb_files_Covid=len(fnmatch.filter(os.listdir(workingpath+covid_files+"/images/"), "*.png"))
all_img_Covid = np.zeros((size_img, size_img, nb_files_Covid),  dtype=np.uint8)
for i in range(0, nb_files_Covid):
    img_Covid = cv2.imread(workingpath+covid_files +"/images/COVID-"+str(i+1)+".png")  
    img_Covid= cv2.cvtColor(img_Covid, cv2.COLOR_BGR2GRAY)
    #img_Covid_mask = cv2.imread(workingpath+covid_files +"/masks/COVID-"+str(i+1)+".png", cv2.IMREAD_GRAYSCALE) # pour lire

    #img_mask_resized = cv2.resize(img_Covid_mask, (img_Covid.shape[0], img_Covid.shape[1]), 0, 0, cv2.INTER_NEAREST)
    #img_Covid_and_mask = cv2.bitwise_and(img_Covid,img_mask_resized)
    #img_Covid_and_mask_resized = cv2.resize(img_Covid_and_mask, (size_img, size_img), 0, 0, cv2.INTER_NEAREST)
    all_img_Covid[:,:,i] = img_Covid
```

```python
nb_files_LungOpacity=len(fnmatch.filter(os.listdir(workingpath+lungOpacity_files+"/images/"), "*.png"))
all_img_LungOpacity = np.zeros((size_img, size_img, nb_files_LungOpacity),  dtype=np.uint8)
for i in range(0, nb_files_LungOpacity):
    img_LungOpacity = cv2.imread(workingpath+lungOpacity_files +"/images/Lung_Opacity-"+str(i+1)+".png")  
    img_LungOpacity = cv2.cvtColor(img_LungOpacity, cv2.COLOR_BGR2GRAY)
    #img_LungOpacity_mask = cv2.imread(workingpath+lungOpacity_files +"/masks/Lung_Opacity-"+str(i+1)+".png", cv2.IMREAD_GRAYSCALE) # pour lire

    #img_mask_resized = cv2.resize(img_LungOpacity_mask, (img_LungOpacity.shape[0], img_LungOpacity.shape[1]), 0, 0, cv2.INTER_NEAREST)
    #img_LungOpacity_and_mask = cv2.bitwise_and(img_LungOpacity,img_mask_resized)
    #img_LungOpacity_and_mask_resized = cv2.resize(img_LungOpacity_and_mask, (size_img, size_img), 0, 0, cv2.INTER_NEAREST)
    all_img_LungOpacity[:,:,i] = img_LungOpacity
```

```python
nb_files_ViralPneumonia=len(fnmatch.filter(os.listdir(workingpath+viralPneumonia_files+"/images/"), "*.png"))
all_img_ViralPneumonia = np.zeros((size_img, size_img, nb_files_ViralPneumonia),  dtype=np.uint8)
for i in range(0, nb_files_ViralPneumonia):
    img_ViralPneumonia = cv2.imread(workingpath+viralPneumonia_files +"/images/Viral Pneumonia-"+str(i+1)+".png")  
    img_ViralPneumonia = cv2.cvtColor(img_ViralPneumonia, cv2.COLOR_BGR2GRAY)
    #img_ViralPneumonia_mask = cv2.imread(workingpath+viralPneumonia_files +"/masks/Viral Pneumonia-"+str(i+1)+".png", cv2.IMREAD_GRAYSCALE) # pour lire

    #img_mask_resized = cv2.resize(img_ViralPneumonia_mask, (img_ViralPneumonia.shape[0], img_ViralPneumonia.shape[1]), 0, 0, cv2.INTER_NEAREST)
    #img_ViralPneumonia_and_mask = cv2.bitwise_and(img_ViralPneumonia,img_mask_resized)
    #img_ViralPneumonia_and_mask_resized = cv2.resize(img_ViralPneumonia_and_mask, (size_img, size_img), 0, 0, cv2.INTER_NEAREST)
    all_img_ViralPneumonia[:,:,i] = img_ViralPneumonia
```

```python
nb_img_normal=all_img_Normal.shape[2]
nb_img_covid=all_img_Covid.shape[2]
nb_img_LO=all_img_LungOpacity.shape[2]
nb_img_VP=all_img_ViralPneumonia.shape[2]

print("Images Normal",nb_img_normal)
print("Images Covid", nb_img_covid)
print("Images Lung Opacity", nb_img_LO)
print("Images Viral Pneumonia", nb_img_VP)

dict={"Normal":nb_img_normal, "Covid":nb_img_covid, "Lung Opacity":nb_files_LungOpacity, "Viral pneumonia":nb_files_ViralPneumonia}

fig3=plt.figure(figsize=(7,7))

plt.barh(dict.keys(), dict.values(), color=["gray", "darkblue", "orange", "red"])

plt.title("Nombre d'images dans les différents dossiers")
plt.show()
```

Il y a un déséquilibre entre les différentes sources de données, les images issues du dossier Normal sont bien plus nombreuses que celles des autres dossiers. Cela interroge sur la manière d'entraîner l'algorithme et sur un possible déséquilibre durant cette phase.

```python
img=all_img_Normal
img_moy=all_img_Normal.mean(axis=-1)
img_std=all_img_Normal.std(axis=-1)
img_med=np.median(all_img_Normal, axis=-1)

imgC=all_img_Covid
imgC_moy=imgC.mean(axis=-1)
imgC_std=all_img_Normal.std(axis=-1)
imgC_med=np.median(all_img_Covid, axis=-1)

imgLO=all_img_LungOpacity
imgLO_moy=imgLO.mean(axis=-1)
imgLO_std=all_img_LungOpacity.std(axis=-1)
imgLO_med=np.median(all_img_LungOpacity, axis=-1)

imgVP=all_img_ViralPneumonia
imgVP_moy=imgVP.mean(axis=-1)
imgVP_std=imgVP.std(axis=-1)
imgVP_med=np.median(all_img_ViralPneumonia, axis=-1)
```

```python

fig2=plt.figure(figsize=(18,13))

ax1=plt.subplot(3,4,1)

plt.imshow(img_moy, cmap="gray", interpolation=None)
plt.title("image moy. Normal")

ax2=plt.subplot(3,4,2)

plt.imshow(imgC_moy, cmap="gray")
plt.title("image moy. Covid")

ax3=plt.subplot(3,4,3)

plt.imshow(imgLO_moy, cmap="gray", interpolation=None)
plt.title("image moy. Lung Opacity")

ax4=plt.subplot(3,4,4)

plt.imshow(imgVP_moy, cmap="gray")
plt.title("image moy. Viral Pneumonia")

ax5=plt.subplot(3,4,5)
plt.imshow(img_std, cmap="gray")
plt.title("std image normale")

ax6=plt.subplot(3,4,6)
plt.imshow(imgC_std, cmap="gray")
plt.title("std image Covid")

ax7=plt.subplot(3,4,7)
plt.imshow(imgLO_std, cmap="gray")
plt.title("std image Lung Opacity")

ax8=plt.subplot(3,4,8)
plt.imshow(imgVP_std, cmap="gray")
plt.title("std image Viral Pneumonia")

ax9=plt.subplot(3,4,9)
plt.imshow(img_med, cmap="gray")
plt.title("médiane image normale")

ax10=plt.subplot(3,4,10)
plt.imshow(imgC_med, cmap="gray")
plt.title("médiane image Covid")

ax11=plt.subplot(3,4,11)
plt.imshow(imgLO_med, cmap="gray")
plt.title("médiane image Lung Opacity")

ax12=plt.subplot(3,4,12)
plt.imshow(imgVP_med, cmap="gray")
plt.title("médiane image Viral Pneumonia")

plt.show()
```

On note des différences de contraste entre les différentes classes du dataset. Les images des patients normaux sont en moyenne plus sombres que celles des patients Covid chez lesquels l'espace pulmonaire apparaît en moyenne plus flou.

L'écart-type indique les variations de pixels pour chacune des classes. On remarque que les zones enregistrant les plus fortes variations (en clair sur les images) se trouvent autour des différentes structures anatomiques (poumons notamment).
