```python
import numpy as np
import cv2 #import OpenCV
import matplotlib.pyplot as plt

workingpath="/home/cyril/liora/projet"
normal_files="/COVID-19_Radiography_Dataset/Normal"
covid_files="/COVID-19_Radiography_Dataset/COVID"
Lung_Opacityfiles="/COVID-19_Radiography_Dataset/Lung_Opacity"
viral_pneumonia_files="/COVID-19_Radiography_Dataset/Viral Pneumonia"

img_normal = cv2.imread(workingpath+normal_files +"/images/Normal-2.png") # pour lire

print('dimension de img :', img_normal.shape)
plt.imshow(img_normal);
plt.xticks([]);
plt.yticks([]);
plt.title("Normal-1.png")
plt.show();




```

```python
img_normal_mask = cv2.imread(workingpath+normal_files +"/masks/Normal-2.png") # pour lire

print('dimension de img_mask :', img_normal_mask.shape)
plt.imshow(img_normal_mask);
plt.xticks([]);
plt.yticks([]);
plt.title("MASK/Normal-2.png")
plt.show();
```

```python

img_mask_resized = cv2.resize(img_normal_mask, (img_normal.shape[0], img_normal.shape[1]), 0, 0, cv2.INTER_NEAREST)
img_normal_and_mask = cv2.bitwise_and(img_normal,img_mask_resized)

plt.imshow(img_normal_and_mask);
plt.xticks([]);
plt.yticks([]);
plt.title("img_normal_and_mask-1")
plt.show();
```
