```python
import numpy as np
import cv2 #import OpenCV
import matplotlib.pyplot as plt

workingpath="/home/cyril/liora/projet"
normal_files="/COVID-19_Radiography_Dataset/Normal"
covid_files="/COVID-19_Radiography_Dataset/COVID"
Lung_Opacityfiles="/COVID-19_Radiography_Dataset/Lung_Opacity"
viral_pneumonia_files="/COVID-19_Radiography_Dataset/Viral Pneumonia"

img_normal = cv2.imread(workingpath+normal_files +"/images/Normal-2.png") # pour lire

print('dimension de img :', img_normal.shape)
plt.imshow(img_normal);
plt.xticks([]);
plt.yticks([]);
plt.title("Normal-1.png")
plt.show();




```

```python

img_normal_gray_red = img_normal[:, :, 0]
img_normal_gray_blue = img_normal[:, :, 1]
img_normal_gray_green = img_normal[:, :, 2]


img_normal_gray_red_minus_blue = img_normal[:, :, 0]
# afficher l'image en niveaux de gris
img_normal_gray_blue_minus_green = img_normal[:, :, 1]
# afficher l'image en niveaux de gris
img_normal_gray_green_minus_blue = img_normal[:, :, 2]

plt.subplot(1, 3, 1)
plt.imshow(img_normal_gray_red, cmap='gray')
plt.xticks([]); plt.yticks([])
plt.title("RED CANAL")

plt.subplot(1, 3, 2)
plt.imshow(img_normal_gray_green, cmap='gray')
plt.xticks([]); plt.yticks([])
plt.title("GREEN CANAL")

plt.subplot(1, 3, 3)
plt.imshow(img_normal_gray_blue, cmap='gray')
plt.xticks([]); plt.yticks([])
plt.title("BLUE CANAL")

plt.subplot(2, 3, 1)
plt.imshow(img_normal_gray_red_minus_blue, cmap='gray')
plt.xticks([]); plt.yticks([])
plt.title("RED CANAL")

plt.subplot(2, 3, 2)
plt.imshow(img_normal_gray_green, cmap='gray')
plt.xticks([]); plt.yticks([])
plt.title("GREEN CANAL")

plt.subplot(2, 3, 3)
plt.imshow(img_normal_gray_blue, cmap='gray')
plt.xticks([]); plt.yticks([])
plt.title("BLUE CANAL")


```
